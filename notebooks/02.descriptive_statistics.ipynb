3. Código en Python para Exportar las Tablas


```python
import pandas as pd
import numpy as np

# 1. Cargar el dataset preprocesado
df = pd.read_csv('../data/breast_cancer_original.csv')

feature_cols = [
    'Clump_thickness', 'Uniformity_of_cell_size', 'Uniformity_of_cell_shape',
    'Marginal_adhesion', 'Single_epithelial_cell_size', 'Bare_nuclei',
    'Bland_chromatin', 'Normal_nucleoli', 'Mitoses'
]

# 2. Asegurar que las variables sean cuantitativas
for col in feature_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# 3. Calcular las estadísticas agregadas por clase
resumen = df.groupby('Class')[feature_cols].agg(['mean', 'std', 'median']).T

# 4. Aplanar el MultiIndex para que cada fila tenga su etiqueta completa
resumen_limpio = resumen.reset_index()

# Renombrar columnas para que sean totalmente claras en el CSV
# (Ajusta 'Benigno_2' y 'Maligno_4' según los valores reales en tu columna Class: 0/1 o 2/4)
resumen_limpio.columns = ['Variable', 'Estadistico', 'Benigno', 'Maligno']

# 5. Redondear los valores a 2 decimales para presentación limpia
resumen_limpio['Benigno'] = resumen_limpio['Benigno'].round(2)
resumen_limpio['Maligno'] = resumen_limpio['Maligno'].round(2)

# 6. Guardar en CSV desactivando el índice numérico automático (index=False)
resumen_limpio.to_csv('../data/estadisticas_descriptivas_cancer.csv', index=False)

print("¡Tabla exportada correctamente!")
print(resumen_limpio.head(9))
```

Dado que la variable "Mitoses" tiene la misma mediana y la diferencia de medias similar en los grupos benignos y malignos, se descartó esta variable.
